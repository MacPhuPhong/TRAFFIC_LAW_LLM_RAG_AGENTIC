# 01 — RQ1: Gemini thuần vs RAG baseline vs Agentic RAG

**Câu hỏi:** Bổ sung RAG rồi Agentic layer đem lại cải thiện bao nhiêu so với Gemini thuần?

**Metric:** F1 (token), ROUGE-L, latency.

**Output:** bảng 3 hàng × 5 cột.

```python
# --- Setup ---
import sys
from pathlib import Path

HERE = Path.cwd()
for p in [HERE] + list(HERE.parents):
    if (p / "source").is_dir() and (p / "research").is_dir():
        TRAFFIC_RAG = p
        break
else:
    raise RuntimeError("Could not locate traffic_rag/ root")

sys.path.insert(0, str(TRAFFIC_RAG))
sys.path.insert(0, str(TRAFFIC_RAG.parent))

from research.utils.langsmith_setup import enable_tracing
enable_tracing(project="traffic-rag-research", run_name="rq1-compare")

EVAL_PATH = TRAFFIC_RAG / "research" / "data" / "eval_qa.jsonl"
RESULTS_DIR = TRAFFIC_RAG / "research" / "results" / "metrics"
FIGURES_DIR = TRAFFIC_RAG / "research" / "results" / "figures"
RESULTS_DIR.mkdir(parents=True, exist_ok=True)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
```

## 1. Chạy 3 pipeline qua eval set

```python
from research.utils.eval_runner import run_over_dataset

for name in ["gemini_only", "vanilla_rag", "agentic_rag"]:
    out = RESULTS_DIR / f"rq1_{name}.jsonl"
    print("==>", name, "→", out)
    run_over_dataset(name, EVAL_PATH, out_path=out, resume=True)
```

## 2. Chấm metric answer-level

```python
from research.utils.metrics import token_f1, rouge_l
from research.utils.eval_runner import load_results
import pandas as pd

rows = []
for name in ["gemini_only", "vanilla_rag", "agentic_rag"]:
    results = load_results(RESULTS_DIR / f"rq1_{name}.jsonl")
    for r in results:
        rows.append({
            "pipeline": name,
            "id": r["id"],
            "f1": token_f1(r.get("answer") or "", r.get("gold_answer") or ""),
            "rouge_l": rouge_l(r.get("answer") or "", r.get("gold_answer") or ""),
            "latency_s": r.get("latency_s"),
        })
df = pd.DataFrame(rows)
summary = df.groupby("pipeline").agg(
    f1=("f1", "mean"),
    rouge_l=("rouge_l", "mean"),
    latency_s=("latency_s", "mean"),
).reset_index()
print(summary)
summary.to_csv(RESULTS_DIR / "rq1_summary.csv", index=False)
```
