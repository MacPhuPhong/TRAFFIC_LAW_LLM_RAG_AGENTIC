## 1. Run Evaluation with Different Prompts

```python
from research.utils.pipelines import PIPELINE_REGISTRY, VanillaRAGPipeline
from research.utils.eval_runner import run_over_dataset
from source.rag_core.generator import LegalAnswerGenerator

# Original Pipeline uses the hardcoded system prompt
out_original = RESULTS_DIR / "rq5_prompt_original.jsonl"
print("==> Running Original Prompt (v5.8.2)...")
run_over_dataset("vanilla_rag", EVAL_PATH, out_path=out_original, resume=True)

# Create Zero-Shot degraded prompt pipeline
class ZeroShotRAGPipeline(VanillaRAGPipeline):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        # Override the generator's prompt to be very basic
        self.generator.system_prompt = "Bạn là AI tư vấn luật. Hãy trả lời câu hỏi dựa trên Context. Nếu không có context, không được tự bịa ra.\nContext:\n{context}"

PIPELINE_REGISTRY["zero_shot_prompt"] = ZeroShotRAGPipeline

out_zeroshot = RESULTS_DIR / "rq5_prompt_zeroshot.jsonl"
print("==> Running Zero-Shot Base Prompt...")
run_over_dataset("zero_shot_prompt", EVAL_PATH, out_path=out_zeroshot, resume=True)
```

## 2. Evaluate Results

```python
import json

def normalize_vn(text):
    if not text: return ""
    text = text.lower()
    text = re.sub(r'[^\w\s]', ' ', text)
    return re.sub(r'\s+', ' ', text).strip()

def token_f1(pred, gold):
    p_t = normalize_vn(pred).split()
    g_t = normalize_vn(gold).split()
    if not p_t or not g_t: return 0.0
    common = Counter(p_t) & Counter(g_t)
    n = sum(common.values())
    if n == 0: return 0.0
    prec = n / len(p_t)
    rec = n / len(g_t)
    return 2 * prec * rec / (prec + rec)

def analyze_f1(path):
    if not path.exists(): return 0.0
    rows = [json.loads(l) for l in open(path)]
    f1s = [token_f1(r.get('answer',''), r.get('gold_answer','')) for r in rows]
    return sum(f1s) / len(f1s) if f1s else 0.0

f1_orig = analyze_f1(out_original)
f1_zero = analyze_f1(out_zeroshot)

print("\n--- PROMPT ABLATION RESULTS ---")
print(f"Original Prompt F1-Score: {f1_orig:.2f}")
print(f"Zero-Shot Prompt F1-Score: {f1_zero:.2f}")
```
